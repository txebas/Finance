Para limpiar estas cadenas y eliminar únicamente los números de línea (que en documentos de la SEC suelen aparecer solos en una línea o como parte de un contador de página), podemos utilizar **expresiones regulares (Regex)**.

El patrón común en tus tres ejemplos es que los números de línea aparecen como elementos aislados dentro de las etiquetas `<div>`.

### Estrategia de Limpieza
Para conservar la estructura pero eliminar los números, debemos identificar:
1.  Números enteros solos: `<div>1</div>`, `<div>2.</div>`.
2.  Secuencias de números de página específicos: `Apple Inc. | 2025 Form 10-K | 1`.
3.  Números con puntos: `48.`.

### Propuesta de Patrón Regex
Puedes utilizar el siguiente patrón para identificar las líneas que quieres **eliminar**:

`^<div align='center'>(?:\d+\.?|.*\|\s\d+)<\/div>$`

---

### Implementación en Python
Aquí tienes un script rápido para procesar tus variables de tipo string:

```python
import re

def limpiar_numeros_linea(texto):
    # Patrón 1: Números solos o con punto final (ej: 48 o 48.)
    # Patrón 2: Formato de Apple (Nombre | Form | Número)
    patron = r"<div align='center'>(?:\d+\.?|\w+\sInc\.\s\|\s\d{4}\sForm\s10-K\s\|\s\d+)</div>"
    
    # Filtramos las líneas que NO coinciden con el patrón de "solo números"
    lineas = texto.split('\n')
    lineas_limpias = [l for l in lineas if not re.fullmatch(patron, l.strip())]
    
    return '\n'.join(lineas_limpias)
```

Basado en tus ejemplos, el resultado filtraría lo siguiente:

| Se ELIMINA (Números de línea/página) | Se CONSERVA (Contenido relevante) |
| :--- | :--- |
| `<div align='center'>2</div>` | `<div align='center'>(Exact name of registrant...)</div>` |
| `<div align='center'>43.</div>` | `<div align='center'>Table of Contents</div>` |
| `<div align='center'>Apple Inc. | 2025 Form 10-K | 1</div>` | `<div align='center'>(In millions, except per share...)</div>` |
| `<div align='center'>56</div>` | `<div align='center'>See accompanying Notes...</div>` |

### Notas adicionales
*   **Caso Apple:** He incluido específicamente el patrón `| 2025 Form 10-K |` porque en ese formato el número de página no viene solo, sino precedido por el nombre de la empresa.
*   **Espacios en blanco:** El script utiliza `.strip()` para evitar errores si hay espacios accidentales al inicio o final de la línea.
*   **Detección de puntos:** He añadido `\.?` para capturar tanto `24` como `24.` que aparecen en tu segundo bloque de ejemplos.

````python
import re
````
````python
from abc import ABC, abstractmethod
from typing import List, Dict, Any

# Librerías necesarias (Instalar si no las tienes: pip install langchain langchain-text-splitters langchain-experimental sentence-transformers)
from langchain_text_splitters import RecursiveCharacterTextSplitter, MarkdownHeaderTextSplitter
from langchain_experimental.text_splitter import SemanticChunker
from langchain_community.embeddings import HuggingFaceEmbeddings

# ==========================================
# 1. INTERFAZ Y ESTRATEGIAS
# ==========================================

class ChunkingStrategy(ABC):
    """Interfaz base para todas las estrategias de chunking."""
    @abstractmethod
    def split(self, text: str) -> List[str]:
        pass

class RecursiveStrategy(ChunkingStrategy):
    """Estrategia estándar: divide por jerarquía de caracteres."""
    def __init__(self, chunk_size=800, chunk_overlap=80):
        self.splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
            separators=["\n\n", "\n", ".", " ", ""]
        )

    def split(self, text: str) -> List[str]:
        return self.splitter.split_text(text)

class SECStructuralStrategy(ChunkingStrategy):
    """Estrategia Estructural: identifica Items específicos de la SEC (10-K)."""
    def __init__(self):
        # Patrón para identificar los 'Items' de un 10-K
        self.item_pattern = r'(ITEM\s+(?:1[A-B]?|[2-9]|10|11|12|13|14|15)\..+?)(?=ITEM\s+(?:1[A-B]?|[2-9]|10|11|12|13|14|15)\.|$)'

    def split(self, text: str) -> List[str]:
        sections = re.findall(self.item_pattern, text, re.IGNORECASE | re.DOTALL)
        return [s.strip() for s in sections] if sections else [text]

class SemanticStrategy(ChunkingStrategy):
    """Estrategia Semántica: divide basándose en cambios de significado."""
    def __init__(self):
        # Usamos un modelo ligero de embeddings local
        embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
        self.splitter = SemanticChunker(embeddings)

    def split(self, text: str) -> List[str]:
        # El SemanticChunker devuelve objetos de documento de LangChain
        docs = self.splitter.create_documents([text])
        return [doc.page_content for doc in docs]

class MarkdownTableStrategy(ChunkingStrategy):
    """Estrategia para tablas: mantiene la estructura Markdown."""
    def __init__(self):
        headers = [("#", "Header 1"), ("##", "Header 2"), ("###", "Header 3")]
        self.splitter = MarkdownHeaderTextSplitter(headers_to_split_on=headers)

    def split(self, text: str) -> List[str]:
        docs = self.splitter.split_text(text)
        return [doc.page_content for doc in docs]

# ==========================================
# 2. ORQUESTADOR (MANAGER)
# ==========================================

class ChunkingManager:
    """Clase para gestionar el cambio de estrategias dinámicamente."""
    def __init__(self, strategy: ChunkingStrategy):
        self._strategy = strategy

    def set_strategy(self, strategy: ChunkingStrategy):
        self._strategy = strategy

    def process_document(self, text: str) -> List[str]:
        print(f"--- Ejecutando: {self._strategy.__class__.__name__} ---")
        chunks = self._strategy.split(text)
        print(f"Total de chunks generados: {len(chunks)}\n")
        return chunks

# ==========================================
# 3. EJECUCIÓN (MAIN)
# ==========================================

if __name__ == "__main__":
    # Simulación de un fragmento de un informe 10-K
    ten_k_content = """
    ITEM 1. BUSINESS. Apple Inc. is a global leader in consumer electronics and software.
    The Company's business strategy leverages its ability to design and develop its own operating systems.
    ITEM 1A. RISK FACTORS. The Company’s business faces various risks, including global economic conditions,
    supply chain disruptions, and intense competition in the smartphone market.
    ITEM 7. MANAGEMENT’S DISCUSSION AND ANALYSIS. This section provides an analysis of the financial results...
    """

    # Inicializar el manager con una estrategia estructural (la ideal para 10-K)
    manager = ChunkingManager(SECStructuralStrategy())

    # 1. Aplicar partición por Items
    items_10k = manager.process_document(ten_k_content)

    # 2. Aplicar partición recursiva sobre el primer Item encontrado (si es muy largo)
    manager.set_strategy(RecursiveStrategy(chunk_size=100, chunk_overlap=20))
    sub_chunks = manager.process_document(items_10k[0])

    # 3. Ejemplo de partición semántica (opcional)
    # Nota: esto descargará un modelo pequeño de embeddings la primera vez.
    # manager.set_strategy(SemanticStrategy())
    # semantic_chunks = manager.process_document(ten_k_content)

    # Visualización rápida del resultado
    for i, chunk in enumerate(sub_chunks[:3]):
        print(f"Chunk {i+1}: {chunk[:80]}...")
        print(f"Chunk {i+1}: {chunk[:]}")
````

````python
import torch
from transformers import AutoTokenizer, AutoModel
import torch.nn.functional as F

# 1. Definir explícitamente las clases de la arquitectura base
# Esto evita que 'transformers' intente adivinar el modelo desde el config.json
tokenizer = AutoTokenizer.from_pretrained('FinanceMTEB/FinE5')
model = AutoModel.from_pretrained('FinanceMTEB/FinE5')

# Función para hacer el "Mean Pooling" (necesario para obtener el vector del texto)
def mean_pooling(model_output, attention_mask):
    token_embeddings = model_output[0]
    input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
    return torch.sum(token_embeddings * input_mask_expanded, 1) / torch.clamp(input_mask_expanded.sum(1), min=1e-9)

# 2. Definir los textos con sus respectivos prefijos
documentos = [
    "passage: El Banco Central Europeo decidió mantener las tasas de interés sin cambios en su última reunión.",
    "passage: Los resultados del trimestre muestran un crecimiento del 5% en los ingresos netos de la compañía.",
    "passage: La inflación subyacente se mantiene estable debido al precio de los servicios."
]
busqueda = "query: ¿Qué pasó con los tipos de interés en Europa?"

todos_los_textos = [busqueda] + documentos

# 3. Tokenizar y pasar el modelo
encoded_input = tokenizer(todos_los_textos, padding=True, truncation=True, return_tensors='pt')

with torch.no_grad():
    model_output = model(**encoded_input)

# 4. Generar los embeddings finales y normalizarlos
embeddings = mean_pooling(model_output, encoded_input['attention_mask'])
embeddings = F.normalize(embeddings, p=2, dim=1)

# Separar la búsqueda de los documentos
embedding_busqueda = embeddings[0]
embeddings_documentos = embeddings[1:]

# 5. Calcular la similitud (al estar normalizados, basta con un producto punto)
similitudes = torch.mm(embedding_busqueda.unsqueeze(0), embeddings_documentos.T).squeeze(0)

# 6. Mostrar resultados
resultados = sorted(zip(documentos, similitudes.tolist()), key=lambda x: x[1], reverse=True)

print(f"Búsqueda: '{busqueda}'\n")
print("Resultados más relevantes:")
for doc, puntaje in resultados:
    print(f"- [Similitud: {puntaje:.4f}] {doc.replace('passage: ', '')}")
````

````python
import re
import tiktoken
import nltk
from abc import ABC, abstractmethod
from typing import List, Union, Any
import torch
from sentence_transformers import SentenceTransformer

# Configuración inicial de NLTK
nltk.download('punkt', quiet=True)

# ==========================================
# 0. Embedding Processor
# ==========================================

class EmbeddingProcessor:
    def __init__(self, model_name: str = "FinanceMTEB/FinE5"):  # all-MiniLM-L6-v2 BAAI/bge-m3 BAAI/bge-small-en-v1.5 BAAI/bge-large-en-v1.5
        self.model = SentenceTransformer(model_name)

    def get_query_embedding(self, query: str) -> torch.Tensor:
        # Ensure output is a PyTorch tensor
        return self.model.encode(query, convert_to_tensor=True)

    def get_document_embeddings(self, documents: List[str]) -> torch.Tensor:
        # Encode documents in batches to optimize performance, ensuring output is a PyTorch tensor
        return self.model.encode(documents, convert_to_tensor=True)

    def compute_similarities(self, query_embedding: torch.Tensor, document_embeddings: torch.Tensor) -> torch.Tensor:
        # Ensure query_embedding has a batch dimension for similarity calculation
        if query_embedding.dim() == 1:
            query_embedding = query_embedding.unsqueeze(0)
        return self.model.similarity(query_embedding, document_embeddings)

# ==========================================
# 1. DEFINICIÓN DE ESTRATEGIAS (MÓDULOS)
# ==========================================

class ChunkingStrategy(ABC):
    @abstractmethod
    def split(self, text: str) -> List[Union[str, Any]]:
        pass

class RecursiveStrategy(ChunkingStrategy):
    """Divide por párrafos y líneas (El estándar de facto)."""
    def __init__(self, size=500, overlap=50):
        from langchain_text_splitters import RecursiveCharacterTextSplitter
        self.splitter = RecursiveCharacterTextSplitter(chunk_size=size, chunk_overlap=overlap)
    def split(self, text: str): return self.splitter.split_text(text)

class TokenStrategy(ChunkingStrategy):
    """Divide según el número de tokens (Ideal para control de costes/contexto)."""
    def __init__(self, size=200, overlap=20):
        from langchain_text_splitters import TokenTextSplitter
        self.splitter = TokenTextSplitter(chunk_size=size, chunk_overlap=overlap)
    def split(self, text: str): return self.splitter.split_text(text)

class MarkdownStrategy(ChunkingStrategy):
    """Divide respetando la jerarquía de encabezados (#, ##, ###)."""
    def __init__(self):
        from langchain_text_splitters import MarkdownHeaderTextSplitter
        # Definimos qué encabezados queremos rastrear
        headers_to_split_on = [
            ("#", "Header 1"),
            ("##", "Header 2"),
            ("###", "Header 3"),
        ]
        self.splitter = MarkdownHeaderTextSplitter(headers_to_split_on=headers_to_split_on)

    def split(self, text: str):
        # Retorna una lista de Documentos con contenido y metadatos del encabezado
        result = self.splitter.split_text(text)
        return [doc.page_content for doc in result]

class SECStructuralStrategy(ChunkingStrategy):
    """Divide por los Items legales del 10-K mediante RegEx."""
    def split(self, text: str):
        # Adjusted regex to correctly capture ITEM X. TITLE pattern consistently
        pattern = r'(ITEM\s+(?:1[A-B]?|[2-9]|1[0-5])(?:[A-C])?\.?\s?[^\n\r]*(?:\n\r|$))(.*?)(?=(?:ITEM\s+(?:1[A-B]?|[2-9]|1[0-5])(?:[A-C])?\.?\s?[^\n\r]*(?:\n\r|$))|\Z)'
        matches = re.findall(pattern, text, re.IGNORECASE | re.DOTALL)

        # The regex now returns (header, content) tuples, we want to combine them
        chunks = []
        for header, content in matches:
            chunks.append((header + content).strip())

        return chunks if chunks else [text]

class SemanticStrategy(ChunkingStrategy):
    """Divide basándose en cambios de significado semántico."""
    def __init__(self):
        from langchain_experimental.text_splitter import SemanticChunker
        from langchain_community.embeddings import HuggingFaceEmbeddings
        # Usamos un modelo local ligero
        self.splitter = SemanticChunker(HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2"))
    def split(self, text: str):
        docs = self.splitter.create_documents([text])
        return [doc.page_content for doc in docs]

# ==========================================
# 2. ORQUESTADOR Y TESTBENCH
# ==========================================

class ChunkingLab:
    def __init__(self, text_sample: str):
        self.text_sample = text_sample
        self.strategies = {
            "Markdown Hierarchical": MarkdownStrategy(),
            "SEC Structural (Regex)": SECStructuralStrategy(),
            "Recursive Character": RecursiveStrategy(size=500, overlap=50),
            "Token Based": TokenStrategy(size=150, overlap=15),
            "Semantic (AI)": SemanticStrategy()
        }
        self.all_chunks = {}
        self.embedding_processor = EmbeddingProcessor()

    def run_all_tests(self):
        print(f"{"="*70}")
        print(f"🧪 COMPARATIVA DE ESTRATEGIAS DE CHUNKING (10-K ANALYSIS)")
        print(f"{"="*70}\n")

        for name, strategy in self.strategies.items():
            try:
                chunks = strategy.split(self.text_sample)
                self.all_chunks[name] = chunks
                print(f"🔹 ESTRATEGIA: {name}")
                print(f"   - Chunks creados: {len(chunks)}")
                if chunks:
                    preview = str(chunks[0]).replace('\n', ' ')[:90]
                    print(f"   - Preview primer chunk: {preview}...")
                print(f"{"-"*70}")
            except Exception as e:
                print(f"❌ Error en {name}: {e}\n")

    def analyze_queries(self, queries: List[str], top_n: int = 3):
        print(f"\n{"="*70}")
        print(f"🔍 ANÁLISIS DE SIMILITUD DE QUERIES CON CHUNKS ")
        print(f"{"="*70}\n")

        for query in queries:
            print(f"➡️ QUERY: \"{query}\"\n")
            query_embedding = self.embedding_processor.get_query_embedding(query)

            for strategy_name, chunks in self.all_chunks.items():
                if not chunks:
                    print(f"  - Estrategia '{strategy_name}': No hay chunks para analizar.")
                    continue

                print(f"  - Estrategia: {strategy_name}")

                # Filter out empty strings before embedding
                non_empty_chunks = [c for c in chunks if c.strip()]
                if not non_empty_chunks:
                    print(f"    (No hay chunks no vacíos para analizar)")
                    continue

                document_embeddings = self.embedding_processor.get_document_embeddings(non_empty_chunks)
                similarities = self.embedding_processor.compute_similarities(query_embedding, document_embeddings)

                # Sort chunks by similarity score
                # similarities is typically a 1xN tensor, so we get the first row
                if similarities.dim() > 1:
                    similarities = similarities.squeeze(0) # Remove batch dimension if present

                sorted_indices = torch.argsort(similarities, descending=True)

                for i in range(min(top_n, len(non_empty_chunks))):
                    idx = sorted_indices[i].item()
                    score = similarities[idx].item()
                    chunk_content = non_empty_chunks[idx].replace('\n', ' ').strip()
                    print(f"    [{i+1}] Score: {score:.4f} -> {chunk_content[:150]}...")
                print(f"    {'--'*30}")
            print(f"\n{'='*60}\n")

# ==========================================
# 3. EJECUCIÓN CON TEXTO DE PRUEBA
# ==========================================

if __name__ == "__main__":
    # Ejemplo de texto en formato Markdown (típico post-procesado de un 10-K)
    # Usamos texto_limpio generado en celdas anteriores
    if 'texto_limpio' not in locals() or not isinstance(texto_limpio, str) or len(texto_limpio) == 0:
        print("Error: 'texto_limpio' variable no disponible o vacía. Asegúrate de ejecutar las celdas anteriores.")
        texto_limpio = "Default content for testing if texto_limpio is missing."

    lab = ChunkingLab(texto_limpio)
    lab.run_all_tests()

    '''
    # Batería de queries para 10-K filings
    investment_queries = [
        "What are the company's primary business segments and products?",
        "What are the key risk factors and uncertainties affecting the business?",
        "Summarize the company's financial performance, including revenue and net income for the last fiscal year.",
        "Who are the principal executive officers and directors?",
        "What are the company's capital expenditures and investment plans?",
        "Describe any significant legal proceedings or contingencies.",
        "What are the company's policies regarding insider trading?"
    ]
    '''

    # Batería de queries ampliada para 10-K filings de empresas tecnológicas
    investment_queries = [
    # --- ESTRATEGIA Y PRODUCTO (BUSINESS & SEGMENTS) ---
    "What are the company's primary business segments and products?",
    "What is the percentage of revenue derived from each reportable business segment?",
    "How does the company describe its competitive position in the cloud computing and software-as-a-service (SaaS) markets?",
    "What are the primary subscription models and how does the company define customer churn or retention?",
    "What is the company's geographic revenue distribution, specifically highlighting exposure to the US, China, and Europe?",
    "Describe the company's strategy regarding the integration of Generative AI into its existing product suite.",
    "What are the key dependencies on third-party intellectual property or open-source software?",

    # --- FACTORES DE RIESGO (RISK FACTORS - ITEM 1A) ---
    "What are the key risk factors and uncertainties affecting the business?",
    "How does the company address risks related to cybersecurity breaches and the protection of user data?",
    "What are the specific risks mentioned regarding global semiconductor supply chains and hardware manufacturing?",
    "How could changes in antitrust laws or government regulations impact the company's market dominance?",
    "What are the risks associated with international operations, including currency fluctuations and geopolitical instability?",
    "Describe the risks of potential service interruptions in data centers or cloud infrastructure.",
    "What impact would a decline in advertising spend have on the company's revenue (if applicable)?",

    # --- RENDIMIENTO FINANCIERO (FINANCIAL PERFORMANCE) ---
    "Summarize the company's financial performance, including revenue and net income for the last fiscal year.",
    "What is the Year-over-Year (YoY) growth rate for the total revenue and the operating margin?",
    "How much did the company spend on Research and Development (R&D) as a percentage of total revenue?",
    "What is the total amount of cash, cash equivalents, and short-term investments held on the balance sheet?",
    "Describe the company's share repurchase (buyback) program and the total amount spent on dividends during the year.",
    "What is the company's effective tax rate and what are the main reasons for its fluctuation?",
    "Provide details on the company's long-term debt, including maturity dates and interest rate exposure.",

    # --- GOBERNANZA Y DIRECTIVA (GOVERNANCE) ---
    "Who are the principal executive officers and directors?",
    "What is the structure of executive compensation, including the balance between salary, bonuses, and stock options?",
    "What are the company's policies regarding insider trading and stock ownership for executives?",
    "Does the company have a dual-class share structure that gives founders or specific individuals disproportionate voting power?",
    "What are the stated ESG (Environmental, Social, and Governance) goals and how is progress being measured?",

    # --- INVERSIÓN Y PROYECTOS FUTUROS (CAPEX & MD&A) ---
    "What are the company's capital expenditures (CapEx) and investment plans for the upcoming fiscal year?",
    "How does management describe the intended use of cash for future acquisitions or strategic partnerships?",
    "What are the company's future unconditional purchase obligations, particularly for data center capacity or hardware components?",
    "Describe any significant investments in 'Other Bets' or non-core business areas mentioned in the MD&A.",

    # --- TEMAS LEGALES Y CONTABLES (LEGAL & ACCOUNTING) ---
    "Describe any significant legal proceedings or contingencies, particularly antitrust or patent infringement lawsuits.",
    "What are the critical accounting estimates that require the most significant management judgment?",
    "Has the company identified any material weaknesses in its internal control over financial reporting?",
    "Describe the impact of recently adopted accounting standards on the company's financial statements.",

    # --- MÉTRICAS ESPECÍFICAS DE TECH ---
    "What is the total headcount (number of employees) and how has it changed compared to the previous year?",
    "How much does the company pay in stock-based compensation (SBC) and what is its impact on GAAP net income?",
    "Does the company mention any reliance on specific AI hardware, such as NVIDIA GPUs or custom-designed chips?",
    "What is the company's backlog or remaining performance obligations (RPO) for long-term contracts?"
    ]

    lab.analyze_queries(investment_queries, top_n=5)

    print("\n" + "="*70)
    print("🎉 TODOS LOS CHUNKS GUARDADOS (PREVIEW DETALLADO) 🎉")
    print(f"{"="*70}\n")

    # This part shows all chunks (truncated) for each strategy, similar to the original request
    for strategy_name, chunks in lab.all_chunks.items():
        print(f"--- Chunks para la Estrategia: {strategy_name} ({len(chunks)} chunks) ---")
        for i, chunk in enumerate(chunks):
            clean_chunk = str(chunk).replace('\n', ' ').strip()
            print(f"  [{i+1}] {clean_chunk[:150]}...")
        print("\n")
````

````python
from sentence_transformers import SentenceTransformer
from typing import List
import torch

class EmbeddingProcessor:
    def __init__(self, model_name: str = "all-MiniLM-L6-v2"):
        self.model = SentenceTransformer(model_name)

    def get_query_embedding(self, query: str) -> torch.Tensor:
        return self.model.encode_query(query)

    def get_document_embeddings(self, documents: List[str]) -> torch.Tensor:
        return self.model.encode_document(documents)

    def compute_similarities(self, query_embedding: torch.Tensor, document_embeddings: torch.Tensor) -> torch.Tensor:
        return self.model.similarity(query_embedding, document_embeddings)

# Instantiate the processor
embedding_processor = EmbeddingProcessor()

# Define your query and documents
query = "What are the risks related to the company's operations?"

# Ensure texto_limpio is a string and split it into a list of documents if it's too long
# For demonstration, I'll split it by lines or use a small portion if it's very large
# Assuming texto_limpio is available from previous cells and is a single large string

# Fallback for documents if texto_limpio is not suitable or available
if 'texto_limpio' in locals() and isinstance(texto_limpio, str) and len(texto_limpio) > 0:
    # Split into sentences or smaller chunks for better embedding results
    import nltk
    #nltk.download('punkt', quiet=True)
    nltk.download('punkt_tab', quiet=True)
    documents = nltk.sent_tokenize(texto_limpio)
    # Limit the number of documents for faster execution if texto_limpio is very large
    if len(documents) > 100:
        documents = documents[:100] # Take first 100 sentences as an example
else:
    documents = [
        "The company faces significant risks from global economic conditions.",
        "Supply chain disruptions can adversely affect our manufacturing process.",
        "Intense competition in the market could impact our revenue growth.",
        "Our innovative products, including smartphones and computers, drive our sales.",
        "Financial performance was strong, with increased net sales and operating income."
    ]


# Get embeddings
query_embeddings = embedding_processor.get_query_embedding(query)
document_embeddings = embedding_processor.get_document_embeddings(documents)

# Print shapes for verification
print(f"Query embedding shape: {query_embeddings.shape}")
print(f"Document embeddings shape: {document_embeddings.shape}")

# Compute similarities
similarities = embedding_processor.compute_similarities(query_embeddings, document_embeddings)
print(f"Similarities: {similarities}")

# Optional: Find the most similar document
most_similar_idx = torch.argmax(similarities)
print(f"\nMost similar document to the query: {documents[most_similar_idx]}")
print(f"Similarity score: {similarities[0, most_similar_idx].item():.4f}")
````
